# Part 1: Continuous Distributions

Continuous distributions model variables that can take any value in a range.

## 1.1 Normal (Gaussian) Distribution

**Parameters:** `mean (μ)`, `stddev (σ)`  
**Use Cases:** Natural measurements, errors, test scores, heights, weights  
**Properties:** Symmetric, bell-shaped, defined by mean and standard deviation

**When to use:**
- Modeling natural phenomena with central tendency
- Sum of many independent random variables (Central Limit Theorem)
- Measurement errors

```typescript
// Create Normal distribution: mean=100, stddev=15 (like IQ scores)
const normal = new NormalDistribution(100, 15);
const rng = new SimpleRNG(42);

// Generate samples
const normalSamples = [];
for (let i = 0; i < 10000; i++) {
  normalSamples.push(normal.sample(rng));
}

console.log('Normal Distribution (μ=100, σ=15):');
console.log(`  Theoretical mean: ${normal.mean}`);
console.log(`  Sample mean: ${mean(normalSamples).toFixed(2)}`);
console.log(`  Theoretical variance: ${normal.variance}`);
console.log(`  Sample std dev: ${standardDeviation(normalSamples).toFixed(2)}`);
console.log(`  PDF at x=100: ${normal.pdf(100).toFixed(6)}`);
console.log(`  CDF at x=115 (1σ above): ${normal.cdf(115).toFixed(4)} (~84.13% expected)`);
```

## 1.2 Uniform Distribution

**Parameters:** `min`, `max`  
**Use Cases:** Complete uncertainty, random selections, round-off errors  
**Properties:** Equal probability for all values in range

**When to use:**
- No information except bounds
- Random number generation
- Maximum entropy (least informative) prior

```typescript
// Create Uniform distribution: min=0, max=100
const uniform = new UniformDistribution(0, 100);
const rng2 = new SimpleRNG(123);

const uniformSamples = [];
for (let i = 0; i < 10000; i++) {
  uniformSamples.push(uniform.sample(rng2));
}

console.log('Uniform Distribution (min=0, max=100):');
console.log(`  Theoretical mean: ${uniform.mean} (should be ${(0+100)/2})`);
console.log(`  Sample mean: ${mean(uniformSamples).toFixed(2)}`);
console.log(`  Theoretical variance: ${uniform.variance.toFixed(2)}`);
console.log(`  Sample std dev: ${standardDeviation(uniformSamples).toFixed(2)}`);
console.log(`  PDF (constant): ${uniform.pdf(50).toFixed(4)}`);
```

## 1.3 Triangular Distribution

**Parameters:** `min`, `mode` (most likely), `max`  
**Use Cases:** Expert estimates with limited data  
**Properties:** Simple three-point estimate, linear slopes

**When to use:**
- Expert judgment (optimistic/pessimistic/most likely)
- Quick approximations
- Project estimation without historical data

```typescript
// Create Triangular distribution: min=50, mode=100, max=200
const triangular = new TriangularDistribution(50, 100, 200);
const rng3 = new SimpleRNG(456);

const triSamples = [];
for (let i = 0; i < 10000; i++) {
  triSamples.push(triangular.sample(rng3));
}

console.log('Triangular Distribution (min=50, mode=100, max=200):');
console.log(`  Theoretical mean: ${triangular.mean.toFixed(2)} (≈ ${(50+100+200)/3})`);
console.log(`  Sample mean: ${mean(triSamples).toFixed(2)}`);
console.log(`  Sample median: ${median(triSamples).toFixed(2)}`);
console.log(`  Sample std dev: ${standardDeviation(triSamples).toFixed(2)}`);
```

## 1.4 LogNormal Distribution

**Parameters:** `mu (μ)`, `sigma (σ)` (of underlying normal)  
**Use Cases:** Asset prices, incomes, file sizes, particle sizes  
**Properties:** Right-skewed, always positive, multiplicative processes

**When to use:**
- Variables that are products of many factors
- Financial returns (prices)
- Growth processes

```typescript
// Create LogNormal distribution: mu=4, sigma=0.5
const lognormal = new LogNormalDistribution(4, 0.5);
const rng4 = new SimpleRNG(789);

const lnSamples = [];
for (let i = 0; i < 10000; i++) {
  lnSamples.push(lognormal.sample(rng4));
}

console.log('LogNormal Distribution (μ=4, σ=0.5):');
console.log(`  Theoretical mean: ${lognormal.mean.toFixed(2)}`);
console.log(`  Sample mean: ${mean(lnSamples).toFixed(2)}`);
console.log(`  Sample median: ${median(lnSamples).toFixed(2)}`);
console.log(`  Sample std dev: ${standardDeviation(lnSamples).toFixed(2)}`);
console.log(`  Note: Right-skewed distribution, mean > median`);
```

## 1.5 Exponential Distribution

**Parameters:** `lambda (λ)` (rate parameter)  
**Use Cases:** Time between events, failure times, service times  
**Properties:** Memoryless, models waiting times

**When to use:**
- Time until next event (Poisson process)
- Reliability analysis (time to failure)
- Queueing theory

```typescript
// Create Exponential distribution: lambda=0.5 (mean time = 2)
const exponential = new ExponentialDistribution(0.5);
const rng5 = new SimpleRNG(111);

const expSamples = [];
for (let i = 0; i < 10000; i++) {
  expSamples.push(exponential.sample(rng5));
}

console.log('Exponential Distribution (λ=0.5):');
console.log(`  Theoretical mean: ${exponential.mean} (= 1/λ = 1/0.5 = 2)`);
console.log(`  Sample mean: ${mean(expSamples).toFixed(2)}`);
console.log(`  Sample median: ${median(expSamples).toFixed(2)}`);
console.log(`  Sample std dev: ${standardDeviation(expSamples).toFixed(2)}`);
console.log(`  Memoryless property: P(X>t+s|X>t) = P(X>s)`);
```

## 1.6 Beta Distribution

**Parameters:** `alpha (α)`, `beta (β)`  
**Use Cases:** Proportions, probabilities, task completion percentages  
**Properties:** Bounded [0,1], flexible shapes

**When to use:**
- Modeling percentages and proportions
- Bayesian prior for probabilities
- PERT distribution foundation

```typescript
// Create Beta distribution: alpha=2, beta=5 (skewed left)
const beta = new BetaDistribution(2, 5);
const rng6 = new SimpleRNG(222);

const betaSamples = [];
for (let i = 0; i < 10000; i++) {
  betaSamples.push(beta.sample(rng6));
}

console.log('Beta Distribution (α=2, β=5):');
console.log(`  Theoretical mean: ${beta.mean.toFixed(4)} (= α/(α+β) = 2/7)`);
console.log(`  Sample mean: ${mean(betaSamples).toFixed(4)}`);
console.log(`  Sample median: ${median(betaSamples).toFixed(4)}`);
console.log(`  Range: [0, 1] - all samples in ${Math.min(...betaSamples).toFixed(4)} to ${Math.max(...betaSamples).toFixed(4)}`);
```

## 1.7 Gamma Distribution

**Parameters:** `shape (k)`, `rate (θ)`  
**Use Cases:** Sum of exponentials, rainfall amounts, insurance claims  
**Properties:** Flexible, generalizes exponential

**When to use:**
- Sum of k independent exponential variables
- Waiting time for k events
- Continuous skewed data

```typescript
// Create Gamma distribution: shape=2, rate=0.5
const gamma = new GammaDistribution(2, 0.5);
const rng7 = new SimpleRNG(333);

const gammaSamples = [];
for (let i = 0; i < 10000; i++) {
  gammaSamples.push(gamma.sample(rng7));
}

console.log('Gamma Distribution (shape=2, rate=0.5):');
console.log(`  Theoretical mean: ${gamma.mean} (= shape/rate = 2/0.5 = 4)`);
console.log(`  Sample mean: ${mean(gammaSamples).toFixed(2)}`);
console.log(`  Sample std dev: ${standardDeviation(gammaSamples).toFixed(2)}`);
console.log(`  Special case: shape=1 → Exponential distribution`);
```

## 1.8 Weibull Distribution

**Parameters:** `shape (k)`, `scale (λ)`  
**Use Cases:** Reliability engineering, failure analysis, wind speeds  
**Properties:** Flexible hazard rate, widely used in survival analysis

**When to use:**
- Product lifetime modeling
- Failure rate analysis (infant mortality, random, wear-out)
- Material strength

```typescript
// Create Weibull distribution: shape=2, scale=100 (Rayleigh)
const weibull = new WeibullDistribution(2, 100);
const rng8 = new SimpleRNG(444);

const weibullSamples = [];
for (let i = 0; i < 10000; i++) {
  weibullSamples.push(weibull.sample(rng8));
}

console.log('Weibull Distribution (shape=2, scale=100):');
console.log(`  Theoretical mean: ${weibull.mean.toFixed(2)}`);
console.log(`  Sample mean: ${mean(weibullSamples).toFixed(2)}`);
console.log(`  Sample std dev: ${standardDeviation(weibullSamples).toFixed(2)}`);
console.log(`  shape=2 is Rayleigh distribution (wind speed modeling)`);
```

## 1.9 Pareto Distribution

**Parameters:** `scale (xₘ)`, `shape (α)`  
**Use Cases:** Wealth distribution, city sizes, 80/20 rule  
**Properties:** Power law, heavy tail

**When to use:**
- Income/wealth inequality
- Network traffic
- Natural disasters severity

```typescript
// Create Pareto distribution: scale=1, shape=2
const pareto = new ParetoDistribution(1, 2);
const rng9 = new SimpleRNG(555);

const paretoSamples = [];
for (let i = 0; i < 10000; i++) {
  paretoSamples.push(pareto.sample(rng9));
}

console.log('Pareto Distribution (scale=1, shape=2):');
console.log(`  Theoretical mean: ${pareto.mean.toFixed(2)}`);
console.log(`  Sample mean: ${mean(paretoSamples).toFixed(2)}`);
console.log(`  Sample median: ${median(paretoSamples).toFixed(2)}`);
console.log(`  80/20 rule: ~80% of values are in lower 20% of range`);
console.log(`  Heavy tail: max sample = ${Math.max(...paretoSamples).toFixed(2)}`);
```

## 1.10 PERT Distribution

**Parameters:** `min`, `mode` (most likely), `max`  
**Use Cases:** Project management, smoother alternative to Triangular  
**Properties:** Based on Beta, smoother than Triangular

**When to use:**
- Project task duration estimates
- Expert judgment with bounded uncertainty
- When smoother distribution than Triangular is desired

```typescript
// Create PERT distribution: min=10, mode=20, max=40
const pert = new PERTDistribution(10, 20, 40);
const rng10 = new SimpleRNG(666);

const pertSamples = [];
for (let i = 0; i < 10000; i++) {
  pertSamples.push(pert.sample(rng10));
}

console.log('PERT Distribution (min=10, mode=20, max=40):');
console.log(`  Theoretical mean: ${pert.mean.toFixed(2)} (weighted: (min+4*mode+max)/6)`);
console.log(`  Sample mean: ${mean(pertSamples).toFixed(2)}`);
console.log(`  Sample median: ${median(pertSamples).toFixed(2)}`);
console.log(`  Smoother peak than Triangular, same three-point estimate`);
```

# Part 2: Discrete Distributions

Discrete distributions model countable events (integers).

## 2.1 Binomial Distribution

**Parameters:** `n` (trials), `p` (success probability)  
**Use Cases:** Number of successes in n independent trials  
**Properties:** Discrete, symmetric when p=0.5

**When to use:**
- Quality control (defects)
- A/B testing (conversions)
- Survey responses (yes/no)

```typescript
// Create Binomial distribution: 20 coin flips, p=0.3
const binomial = new BinomialDistribution(20, 0.3);
const rng11 = new SimpleRNG(777);

const binomialSamples = [];
for (let i = 0; i < 10000; i++) {
  binomialSamples.push(binomial.sample(rng11));
}

console.log('Binomial Distribution (n=20, p=0.3):');
console.log(`  Theoretical mean: ${binomial.mean} (= n*p = 20*0.3 = 6)`);
console.log(`  Sample mean: ${mean(binomialSamples).toFixed(2)}`);
console.log(`  Theoretical variance: ${binomial.variance} (= n*p*(1-p))`);
console.log(`  PMF at k=6: ${binomial.pmf(6).toFixed(4)}`);
console.log(`  Range: integers from 0 to ${binomial.n}`);
```

## 2.2 Poisson Distribution

**Parameters:** `lambda (λ)` (average rate)  
**Use Cases:** Number of events in fixed time/space  
**Properties:** Rare events, approximates Binomial for large n, small p

**When to use:**
- Customer arrivals per hour
- Defects per unit
- Website hits per minute

```typescript
// Create Poisson distribution: lambda=5 (average 5 events)
const poisson = new PoissonDistribution(5);
const rng12 = new SimpleRNG(888);

const poissonSamples = [];
for (let i = 0; i < 10000; i++) {
  poissonSamples.push(poisson.sample(rng12));
}

console.log('Poisson Distribution (λ=5):');
console.log(`  Theoretical mean: ${poisson.mean} (= λ)`);
console.log(`  Sample mean: ${mean(poissonSamples).toFixed(2)}`);
console.log(`  Theoretical variance: ${poisson.variance} (= λ, mean=variance)`);
console.log(`  PMF at k=5: ${poisson.pmf(5).toFixed(4)}`);
console.log(`  Example: Average 5 customers/hour, what's P(exactly 5)? ${(poisson.pmf(5)*100).toFixed(1)}%`);
```

## 2.3 Geometric Distribution

**Parameters:** `p` (success probability)  
**Use Cases:** Trials until first success  
**Properties:** Memoryless (discrete), models waiting time

**When to use:**
- Number of attempts until success
- Quality control inspection
- Search problems

```typescript
// Create Geometric distribution: p=0.2 (20% success rate)
const geometric = new GeometricDistribution(0.2);
const rng13 = new SimpleRNG(999);

const geometricSamples = [];
for (let i = 0; i < 10000; i++) {
  geometricSamples.push(geometric.sample(rng13));
}

console.log('Geometric Distribution (p=0.2):');
console.log(`  Theoretical mean: ${geometric.mean} (= 1/p = 1/0.2 = 5)`);
console.log(`  Sample mean: ${mean(geometricSamples).toFixed(2)}`);
console.log(`  Theoretical variance: ${geometric.variance.toFixed(2)}`);
console.log(`  PMF at k=1 (first try): ${geometric.pmf(1).toFixed(4)} (= p = 0.2)`);
console.log(`  Memoryless: P(X>m+n|X>m) = P(X>n)`);
```

## 2.4 Hypergeometric Distribution

**Parameters:** `N` (population), `K` (successes in pop), `n` (draws)  
**Use Cases:** Sampling WITHOUT replacement  
**Properties:** Like Binomial but without replacement

**When to use:**
- Card games (dealing hands)
- Quality control (finite lots)
- Lottery probabilities

```typescript
// Create Hypergeometric: 50 items, 20 defective, draw 10
const hypergeometric = new HypergeometricDistribution(50, 20, 10);
const rng14 = new SimpleRNG(1010);

const hyperSamples = [];
for (let i = 0; i < 10000; i++) {
  hyperSamples.push(hypergeometric.sample(rng14));
}

console.log('Hypergeometric Distribution (N=50, K=20, n=10):');
console.log(`  Population: 50 items, 20 defective (40% defect rate)`);
console.log(`  Drawing: 10 items without replacement`);
console.log(`  Theoretical mean: ${hypergeometric.mean} (≈ n*K/N = 10*20/50 = 4)`);
console.log(`  Sample mean: ${mean(hyperSamples).toFixed(2)}`);
console.log(`  PMF at k=4: ${hypergeometric.pmf(4).toFixed(4)}`);
console.log(`  Without replacement → dependencies between draws`);
```
