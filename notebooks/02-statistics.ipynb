## Setup and Imports

```typescript
// Import Argo statistical functions
const argo = require('../packages/argo-core/dist/index');

const {
  // Distributions (for generating sample data)
  NormalDistribution,
  UniformDistribution,
  ExponentialDistribution,
  SimpleRNG,
  
  // Descriptive Statistics
  mean,
  median,
  mode,
  min,
  max,
  range,
  variance,
  standardDeviation,
  geometricMean,
  harmonicMean,
  skewness,
  kurtosis,
  
  // Percentiles & Quantiles
  percentile,
  quantile,
  quartiles,
  iqr,
  percentiles,
  
  // Confidence Intervals
  confidenceIntervalNormal,
  confidenceIntervalBootstrap,
  marginOfError,
  sampleSize,
  
  // Distribution Fitting
  fitNormal,
  fitLogNormal,
  goodnessOfFit
} = argo;

console.log('✅ Statistical functions loaded successfully!');
console.log('Total functions available: 30');
```

## Generate Sample Data

Let's create sample datasets to demonstrate the statistical functions.

```typescript
// Generate normal distribution samples (student test scores)
const rng1 = new SimpleRNG(42);
const normalDist = new NormalDistribution(75, 10);
const testScores = [];
for (let i = 0; i < 1000; i++) {
  testScores.push(normalDist.sample(rng1));
}

// Generate skewed data (income distribution)
const rng2 = new SimpleRNG(123);
const expDist = new ExponentialDistribution(0.00002);
const incomes = [];
for (let i = 0; i < 1000; i++) {
  incomes.push(expDist.sample(rng2));
}

// Small sample for confidence intervals
const smallSample = testScores.slice(0, 30);

console.log('Sample datasets created:');
console.log(`  Test scores (normal): ${testScores.length} samples`);
console.log(`  Incomes (skewed): ${incomes.length} samples`);
console.log(`  Small sample: ${smallSample.length} samples`);
```

# Part 1: Descriptive Statistics (12 functions)

Summarize and describe the characteristics of data.

## 1.1 Central Tendency (5 functions)

Measures of the "center" or "typical value" of a dataset.

```typescript
console.log('=== Central Tendency Measures ===\n');

// 1. Mean (arithmetic average)
const avgScore = mean(testScores);
console.log(`1. Mean (average):`);
console.log(`   Test scores: ${avgScore.toFixed(2)}`);
console.log(`   Use: General "typical" value, sensitive to outliers\n`);

// 2. Median (middle value)
const medianScore = median(testScores);
const medianIncome = median(incomes);
console.log(`2. Median (middle value):`);
console.log(`   Test scores: ${medianScore.toFixed(2)}`);
console.log(`   Incomes: $${medianIncome.toFixed(0)}`);
console.log(`   Use: Robust to outliers, better for skewed data\n`);

// 3. Mode (most frequent value)
const dataWithMode = [1, 2, 2, 3, 3, 3, 4, 4, 5];
const modes = mode(dataWithMode);
console.log(`3. Mode (most frequent):`);
console.log(`   Data: [${dataWithMode.join(', ')}]`);
console.log(`   Mode(s): [${modes.join(', ')}]`);
console.log(`   Use: Categorical data, finding most common value\n`);

// 4. Geometric Mean (nth root of product)
const growthRates = [1.05, 1.08, 1.03, 1.06]; // 5%, 8%, 3%, 6% growth
const avgGrowth = geometricMean(growthRates);
console.log(`4. Geometric Mean:`);
console.log(`   Growth rates: [${growthRates.map(r => ((r-1)*100).toFixed(0) + '%').join(', ')}]`);
console.log(`   Average growth: ${((avgGrowth-1)*100).toFixed(2)}%`);
console.log(`   Use: Rates, ratios, multiplicative processes\n`);

// 5. Harmonic Mean (reciprocal of mean of reciprocals)
const speeds = [60, 40, 30]; // mph in different segments
const avgSpeed = harmonicMean(speeds);
console.log(`5. Harmonic Mean:`);
console.log(`   Speeds: [${speeds.join(', ')}] mph`);
console.log(`   Average speed: ${avgSpeed.toFixed(2)} mph`);
console.log(`   Use: Rates (speed, density), averaging ratios`);
```

## 1.2 Dispersion (5 functions)

Measures of how spread out the data is.

```typescript
console.log('\n=== Dispersion Measures ===\n');

// 1. Min & Max
const minScore = min(testScores);
const maxScore = max(testScores);
console.log(`1. Minimum: ${minScore.toFixed(2)}`);
console.log(`2. Maximum: ${maxScore.toFixed(2)}`);
console.log(`   Use: Find extremes, detect outliers\n`);

// 3. Range (max - min)
const scoreRange = range(testScores);
console.log(`3. Range (max - min): ${scoreRange.toFixed(2)}`);
console.log(`   Use: Quick measure of spread, sensitive to outliers\n`);

// 4. Variance (average squared deviation)
const scoreVariance = variance(testScores);
console.log(`4. Variance: ${scoreVariance.toFixed(2)}`);
console.log(`   Use: Foundation for many statistics, in squared units\n`);

// 5. Standard Deviation (square root of variance)
const scoreStdDev = standardDeviation(testScores);
console.log(`5. Standard Deviation: ${scoreStdDev.toFixed(2)}`);
console.log(`   Use: Most common spread measure, same units as data`);
console.log(`   68% of data within 1 SD: [${(avgScore - scoreStdDev).toFixed(1)}, ${(avgScore + scoreStdDev).toFixed(1)}]`);
```

## 1.3 Shape (2 functions)

Measures of the distribution's shape.

```typescript
console.log('\n=== Shape Measures ===\n');

// 1. Skewness (asymmetry)
const scoreSkew = skewness(testScores);
const incomeSkew = skewness(incomes);
console.log(`1. Skewness (asymmetry):`);
console.log(`   Test scores: ${scoreSkew.toFixed(3)} (near 0 = symmetric)`);
console.log(`   Incomes: ${incomeSkew.toFixed(3)} (>0 = right-skewed)`);
console.log(`   Interpretation:`);
console.log(`   - skew = 0: symmetric (normal distribution)`);
console.log(`   - skew > 0: right tail longer (income, wealth)`);
console.log(`   - skew < 0: left tail longer\n`);

// 2. Kurtosis (tail heaviness)
const scoreKurt = kurtosis(testScores);
const incomeKurt = kurtosis(incomes);
console.log(`2. Kurtosis (tail heaviness):`);
console.log(`   Test scores: ${scoreKurt.toFixed(3)} (near 0 = normal)`);
console.log(`   Incomes: ${incomeKurt.toFixed(3)} (>0 = heavy tails)`);
console.log(`   Interpretation (Fisher's excess kurtosis):`);
console.log(`   - kurtosis = 0: normal distribution`);
console.log(`   - kurtosis > 0: heavy tails, more outliers`);
console.log(`   - kurtosis < 0: light tails, fewer outliers`);
```

# Part 2: Percentiles & Quantiles (5 functions)

Divide data into parts and find specific cutoff values.

```typescript
console.log('\n=== Percentiles & Quantiles ===\n');

// 1. Percentile (find value at specific percentage)
const p50 = percentile(testScores, 50);
const p90 = percentile(testScores, 90);
const p95 = percentile(testScores, 95);
console.log(`1. Percentile (value at X%):`);
console.log(`   50th percentile: ${p50.toFixed(2)} (median)`);
console.log(`   90th percentile: ${p90.toFixed(2)} (top 10%)`);
console.log(`   95th percentile: ${p95.toFixed(2)} (top 5%)`);
console.log(`   Use: Grade cutoffs, performance benchmarks\n`);

// 2. Quantile (same as percentile but 0-1 scale)
const q25 = quantile(testScores, 0.25);
const q75 = quantile(testScores, 0.75);
console.log(`2. Quantile (value at proportion):`);
console.log(`   0.25 quantile: ${q25.toFixed(2)} (25th percentile)`);
console.log(`   0.75 quantile: ${q75.toFixed(2)} (75th percentile)`);
console.log(`   Use: Same as percentile, different scale\n`);

// 3. Quartiles (divide into 4 parts)
const q = quartiles(testScores);
console.log(`3. Quartiles (Q1, Q2, Q3):`);
console.log(`   Q1 (25%): ${q.q1.toFixed(2)}`);
console.log(`   Q2 (50%): ${q.q2.toFixed(2)} (median)`);
console.log(`   Q3 (75%): ${q.q3.toFixed(2)}`);
console.log(`   Use: Box plots, outlier detection\n`);

// 4. IQR (Interquartile Range = Q3 - Q1)
const iqrValue = iqr(testScores);
console.log(`4. IQR (Interquartile Range):`);
console.log(`   IQR = Q3 - Q1 = ${q.q3.toFixed(2)} - ${q.q1.toFixed(2)} = ${iqrValue.toFixed(2)}`);
console.log(`   Middle 50% of data spread over ${iqrValue.toFixed(2)} points`);
console.log(`   Use: Robust spread measure, outlier detection\n`);

// 5. Percentiles (multiple at once)
const pctls = percentiles(testScores, [10, 25, 50, 75, 90]);
console.log(`5. Percentiles (batch calculation):`);
console.log(`   10th: ${pctls[0].toFixed(2)}`);
console.log(`   25th: ${pctls[1].toFixed(2)}`);
console.log(`   50th: ${pctls[2].toFixed(2)}`);
console.log(`   75th: ${pctls[3].toFixed(2)}`);
console.log(`   90th: ${pctls[4].toFixed(2)}`);
console.log(`   Use: Efficient calculation of multiple percentiles`);
```

# Part 3: Confidence Intervals (4 functions)

Estimate population parameters from sample data.

```typescript
console.log('\n=== Confidence Intervals ===\n');

// 1. Confidence Interval (Normal/t-distribution)
const ci95 = confidenceIntervalNormal(smallSample, 0.95);
const ci99 = confidenceIntervalNormal(smallSample, 0.99);
console.log(`1. Confidence Interval (Parametric):`);
console.log(`   Sample size: ${smallSample.length}`);
console.log(`   Sample mean: ${mean(smallSample).toFixed(2)}`);
console.log(`   95% CI: [${ci95.lower.toFixed(2)}, ${ci95.upper.toFixed(2)}]`);
console.log(`   99% CI: [${ci99.lower.toFixed(2)}, ${ci99.upper.toFixed(2)}] (wider)`);
console.log(`   Use: Estimate population mean from sample\n`);

// 2. Confidence Interval (Bootstrap - non-parametric)
const rngBoot = new SimpleRNG(999);
const ciBoot = confidenceIntervalBootstrap(smallSample, 0.95, 1000, rngBoot);
console.log(`2. Confidence Interval (Bootstrap):`);
console.log(`   95% Bootstrap CI: [${ciBoot.lower.toFixed(2)}, ${ciBoot.upper.toFixed(2)}]`);
console.log(`   Use: When normality assumption doesn't hold\n`);

// 3. Margin of Error
const moe95 = marginOfError(smallSample, 0.95);
const moe99 = marginOfError(smallSample, 0.99);
console.log(`3. Margin of Error:`);
console.log(`   95% confidence: ±${moe95.toFixed(2)}`);
console.log(`   99% confidence: ±${moe99.toFixed(2)}`);
console.log(`   Report: ${mean(smallSample).toFixed(2)} ± ${moe95.toFixed(2)}`);
console.log(`   Use: Polls, surveys, scientific measurements\n`);

// 4. Sample Size (required for desired margin)
const stdDev = standardDeviation(smallSample);
const n1 = sampleSize(1, stdDev, 0.95);
const n2 = sampleSize(2, stdDev, 0.95);
console.log(`4. Sample Size Calculation:`);
console.log(`   For margin of error = 1: need ${n1} samples`);
console.log(`   For margin of error = 2: need ${n2} samples`);
console.log(`   Use: Study planning, survey design`);
```

# Part 4: Distribution Fitting (3 functions)

Fit probability distributions to observed data.

```typescript
console.log('\n=== Distribution Fitting ===\n');

// 1. Fit Normal Distribution (MLE)
const normalFit = fitNormal(testScores);
console.log(`1. Fit Normal Distribution:`);
console.log(`   Estimated μ (mu): ${normalFit.mu.toFixed(2)}`);
console.log(`   Estimated σ (sigma): ${normalFit.sigma.toFixed(2)}`);
console.log(`   Original distribution: μ=75, σ=10`);
console.log(`   Use: Model normal data, create distribution from samples\n`);

// 2. Fit LogNormal Distribution
const lognormalFit = fitLogNormal(incomes);
console.log(`2. Fit LogNormal Distribution:`);
console.log(`   Estimated μ (log scale): ${lognormalFit.mu.toFixed(2)}`);
console.log(`   Estimated σ (log scale): ${lognormalFit.sigma.toFixed(2)}`);
console.log(`   Use: Income, wealth, prices (right-skewed data)\n`);

// 3. Goodness of Fit (Kolmogorov-Smirnov test)
const fittedDist = new NormalDistribution(normalFit.mu, normalFit.sigma);
const rngGOF = new SimpleRNG(777);
const gofResult = goodnessOfFit(testScores, fittedDist, rngGOF, 1000);
console.log(`3. Goodness of Fit Test:`);
console.log(`   KS statistic: ${gofResult.ksStatistic.toFixed(6)}`);
console.log(`   p-value: ${gofResult.pValue.toFixed(4)}`);
if (gofResult.pValue > 0.05) {
  console.log(`   ✓ Data fits normal distribution (p > 0.05)`);
} else {
  console.log(`   ✗ Data doesn't fit normal distribution (p < 0.05)`);
}
console.log(`   Use: Validate distribution assumptions`);
```

# Part 5: Real-World Example

## Manufacturing Quality Control Analysis

```typescript
// Generate widget weights (target: 100g, some variation)
const rngQC = new SimpleRNG(888);
const widgetDist = new NormalDistribution(100.2, 2.5);
const widgetWeights = [];
for (let i = 0; i < 500; i++) {
  widgetWeights.push(widgetDist.sample(rngQC));
}

console.log('=== Quality Control Report ===\n');

// Descriptive statistics
console.log('Descriptive Statistics:');
console.log(`  Sample size: ${widgetWeights.length}`);
console.log(`  Mean weight: ${mean(widgetWeights).toFixed(2)}g (target: 100g)`);
console.log(`  Median: ${median(widgetWeights).toFixed(2)}g`);
console.log(`  Std deviation: ${standardDeviation(widgetWeights).toFixed(2)}g`);
console.log(`  Range: [${min(widgetWeights).toFixed(2)}, ${max(widgetWeights).toFixed(2)}]g\n`);

// Quality benchmarks
const q = quartiles(widgetWeights);
console.log('Quality Benchmarks:');
console.log(`  25th percentile: ${q.q1.toFixed(2)}g`);
console.log(`  50th percentile: ${q.q2.toFixed(2)}g`);
console.log(`  75th percentile: ${q.q3.toFixed(2)}g`);
console.log(`  IQR: ${iqr(widgetWeights).toFixed(2)}g\n`);

// Outlier detection (1.5 × IQR rule)
const iqrVal = iqr(widgetWeights);
const lowerFence = q.q1 - 1.5 * iqrVal;
const upperFence = q.q3 + 1.5 * iqrVal;
const outliers = widgetWeights.filter(w => w < lowerFence || w > upperFence);
console.log('Outlier Detection:');
console.log(`  Lower fence: ${lowerFence.toFixed(2)}g`);
console.log(`  Upper fence: ${upperFence.toFixed(2)}g`);
console.log(`  Outliers found: ${outliers.length} (${(outliers.length/widgetWeights.length*100).toFixed(1)}%)\n`);

// Process capability
const ci = confidenceIntervalNormal(widgetWeights, 0.95);
console.log('Process Capability:');
console.log(`  95% CI: [${ci.lower.toFixed(2)}, ${ci.upper.toFixed(2)}]g`);
console.log(`  Margin of error: ±${marginOfError(widgetWeights, 0.95).toFixed(2)}g\n`);

// Distribution fit
const fit = fitNormal(widgetWeights);
console.log('Distribution Fit:');
console.log(`  Fitted Normal(μ=${fit.mu.toFixed(2)}, σ=${fit.sigma.toFixed(2)})`);
console.log(`  Skewness: ${skewness(widgetWeights).toFixed(3)} (near 0 = symmetric)`);
console.log(`  Kurtosis: ${kurtosis(widgetWeights).toFixed(3)} (near 0 = normal)`);

console.log('\n✅ Quality control analysis complete!');
```
